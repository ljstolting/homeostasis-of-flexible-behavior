```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*np.pi,1000)
# y = .6*np.sin(x)*(1-0.5*np.sin(x))
y = -np.e**-np.sin(x)
plt.plot(x,y,color='k')
b = np.average(y)
plt.hlines(b,0,2*np.pi,linewidth=3)
plt.fill_between(x,y,b,alpha = .5)
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
x = np.linspace(0,2*np.pi,1000)
y = np.sin(x)
plt.plot(x,y,color='k')
plt.hlines([-.7],0,2*np.pi)
plt.fill_between(x[np.where(y<-.7)],-.7,y[np.where(y<-.7)],alpha = .9)
for ub in [-.2,.2,.5,.7]:
    plt.hlines(ub,0,2*np.pi,color='tab:orange')
    plt.fill_between(x[np.where(y>ub)],ub,y[np.where(y>ub)],color='tab:orange',alpha=.5)
plt.show()
```
