make up for my mistake in not printing the best individuals to files by reading in the last line of the evol files
and for not storing the best fitnesses in a readable format by storing them in a single file

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
forward_fitnesses = np.zeros(100)
for i in range(100):
    with open(f'./Forward Walkers/{i}/evol.dat') as f:
        lines=f.readlines()
        phenotype = lines[-1]
        phenotype = phenotype.split()
        stats = lines[-2]
        stats = stats.split()
        forward_fitnesses[i] = float(stats[1])

    with open(f'./Forward Walkers/{i}/bestind.dat', 'w') as f:
        f.write('2'+'\n\n')
        f.write(" ".join(phenotype[0:2]) + '\n\n')
        f.write(" ".join(phenotype[2:4]) + '\n\n')
        f.write('1 1' + '\n\n')
        f.write(" ".join(phenotype[4:6]) + '\n')
        f.write(" ".join(phenotype[6:8]))
```

```python
np.savetxt('./Forward Walkers/fitnesses.dat', forward_fitnesses,fmt='%.5f',delimiter='\n')
```

```python
backward_fitnesses = np.zeros(100)
for i in range(100):
    with open(f'./Backward Walkers/{i}/evol.dat') as f:
        lines=f.readlines()
        phenotype = lines[-1]
        phenotype = phenotype.split()
        stats = lines[-2]
        stats = stats.split()
        backward_fitnesses[i] = float(stats[1])

    with open(f'./Backward Walkers/{i}/bestind.dat', 'w') as f:
        f.write('2'+'\n\n')
        f.write(" ".join(phenotype[0:2]) + '\n\n')
        f.write(" ".join(phenotype[2:4]) + '\n\n')
        f.write('1 1' + '\n\n')
        f.write(" ".join(phenotype[4:6]) + '\n')
        f.write(" ".join(phenotype[6:8]))
```

```python
#note that these were evolved without the ability to use the coast
#will test whether the coast changes their fitness at all, and will re-evolve in future no matter what
np.savetxt('./Backward Walkers/backward_fitnesses.dat', backward_fitnesses,fmt='%.5f',delimiter='\n')
```
