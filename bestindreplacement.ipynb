make up for my mistake in not printing the best individuals to files by reading in the last line of the evol files

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
for i in range(100):
    with open(f'./Forward Walkers/{i}/evol.dat') as f:
        lines=f.readlines()
        phenotype = lines[-1]
        phenotype = phenotype.split()

    with open(f'./Forward Walkers/{i}/bestind.dat', 'w') as f:
        f.write('2'+'\n\n')
        f.write(" ".join(phenotype[0:2]) + '\n\n')
        f.write(" ".join(phenotype[2:4]) + '\n\n')
        f.write('1 1' + '\n\n')
        f.write(" ".join(phenotype[4:6]) + '\n')
        f.write(" ".join(phenotype[6:8]))
```
